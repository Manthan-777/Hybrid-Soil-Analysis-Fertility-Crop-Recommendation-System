```python
s_ndf = s_ndf.drop(columns = ['Name', 'Yield', 'K_Ratio', 'P_Ratio', 'N_Ratio', 'Season'])
```

```python
Y = s_ndf['Fertility']
Y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

```python
print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix
import joblib
# 1. Identify categorical and numerical features
categorical_cols = ['Soil_Type', 'Photoperiod']
numerical_cols = [col for col in X_train.columns if col not in categorical_cols]

# 2. Build preprocessing transformer for categorical variables
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ],
    remainder='passthrough'
)

# 3. Define pipeline with balanced class weights
fertility_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced'))
])

# 4. Train the model
fertility_model.fit(X_train, Y_train)
print("Fertility Classifier trained successfully!")

# 5. Evaluate on test set
Y_pred = fertility_model.predict(X_test)

print("\n--- Classification Report ---")
print(classification_report(Y_test, Y_pred))

print("--- Confusion Matrix ---")
print(confusion_matrix(Y_test, Y_pred))
```

```python
joblib.dump(fertility_model, 'fertility_classifier_model.pkl')
print("Saved fertility_classifier_model.pkl successfully!")
```

```python
# Load all 3 trained models
crop_model = joblib.load('crop_recommendation_model.pkl')
macro_fertility_model = joblib.load('fertility_classifier_model.pkl')
micro_fertility_model = joblib.load('soil_micro_fertility_model.pkl')

fertility_labels = {0: 'Low', 1: 'Moderate', 2: 'High'}

def unified_agricultural_advisor_v2(
    # Core Nutrients & Soil Chemistry
    n, p, k, ph,
    # Environmental & Light Parameters
    temp, humidity, rainfall, soil_type, photoperiod, light_hours, light_intensity, rh,
    # Micronutrients & Conductance
    ec, oc, s, zn, fe, cu, mn, b
):
    # 1. Feature DataFrame for Model 1 (Crop Recommendation)
    crop_input = pd.DataFrame([{
        'N': n, 'P': p, 'K': k,
        'temperature': temp, 'humidity': humidity, 'ph': ph, 'rainfall': rainfall
    }])

    # 2. Feature DataFrame for Model 2 (Macro Soil Fertility)
    macro_input = pd.DataFrame([{
        'Nitrogen': n, 'Phosphorus': p, 'Potassium': k, 'pH': ph,
        'Temperature': temp, 'Rainfall': rainfall, 'Soil_Type': soil_type,
        'Photoperiod': photoperiod, 'Light_Hours': light_hours,
        'Light_Intensity': light_intensity, 'Rh': rh
    }])

    # 3. Feature DataFrame for Model 3 (Micro & Biochemical Fertility)
    micro_input = pd.DataFrame([{
        'N': n, 'P': p, 'K': k, 'pH': ph,
        'EC': ec, 'OC': oc, 'S': s, 'Zn': zn,
        'Fe': fe, 'Cu': cu, 'Mn': mn, 'B': b
    }])

    # Run Predictions
    predicted_crop = crop_model.predict(crop_input)[0]
    macro_fertility_pred = macro_fertility_model.predict(macro_input)[0]
    micro_fertility_pred = micro_fertility_model.predict(micro_input)[0]

    # Map label if model outputs numeric ID
    micro_class = fertility_labels.get(micro_fertility_pred, micro_fertility_pred)

    return {
        'Recommended Crop': str(predicted_crop).capitalize(),
        'Macro Soil Fertility': macro_fertility_pred,
        'Micro & Biochemical Soil Fertility': micro_class
    }

```

```python
sample = {
    # Core Nutrients & Soil Chemistry
    'n': 320,           # High Nitrogen (Meets Class 2 threshold)
    'p': 35,            # High Phosphorus
    'k': 520,           # High Potassium
    'ph': 6.6,          # Optimal Range

    # Climate & Environmental
    'temp': 40.0,
    'humidity': 85.0,
    'rainfall': 220.0,
    'soil_type': 'Sandy',
    'photoperiod': 'Day Neutral',
    'light_hours': 12.0,
    'light_intensity': 600.0,
    'rh': 85.0,

    # Micronutrients & Conductance
    'ec': 0.55,         # Electrical Conductivity (dS/m)
    'oc': 0.70,         # High Organic Carbon (%)
    's': 8.5,           # Sulphur (ppm)
    'zn': 0.45,         # Zinc (ppm)
    'fe': 4.10,         # Iron (ppm)
    'cu': 0.85,         # Copper (ppm)
    'mn': 8.90,         # Manganese (ppm)
    'b': 0.65           # Boron (ppm)
}

# Run prediction using the tri-model advisor function
result = unified_agricultural_advisor_v2(**sample)
print(result)
```
